```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_2019 = pd.read_csv("data/PropertyValue/2019-property-sales-data.csv")

print("2019 Property Sales Data:")
print(df_2019.head())
```

```python
df_2020 = pd.read_csv("data/PropertyValue/armslengthsales_2020_valid.csv")

print("2020 Property Sales Data:")
print(df_2020.head)
```

```python
df_2021 = pd.read_csv("data/PropertyValue/armslengthsales_2021_valid.csv")

print("2021 Property Sales Data:")
print(df_2021.head)
```

```python
df_2022 = pd.read_csv("data/PropertyValue/armslengthsales_2022_valid_20230404.csv")

print("2022 Property Sales Data:")
print(df_2022.head)
```

```python
df_2023 = pd.read_csv("data/PropertyValue/armslengthsales_2023_valid.csv")

print("2023 Property Sales Data:")
print(df_2023.head)
```

```python
df_2024 = pd.read_csv("data/PropertyValue/armslengthsales_2024_valid_20250917.csv")

print("2024 Property Sales Data:")
print(df_2024.head)
# Keep only rows where SoldAsVacantLand is not equal to "Y" and drop the SoldAsVacantLand column
df_2024 = df_2024[
    df_2024["SoldAsVacantLand"]
      .astype(str).str.strip().str.upper() != "Y"
]

df_2024 = df_2024.drop(columns=["SoldAsVacantLand"])
```

```python
df_2025 = pd.read_csv("data/PropertyValue/armslengthsales_2025_valid_20261019.csv")

print("2025 Property Sales Data:")
print(df_2025.head)
# Keep only rows where SoldasVacantFlag is not equal to 1 and drop the SoldasVacantFlag column
df_2025 = df_2025[
    df_2025["SoldasVacantFlag"].astype(str).str.strip() != "1"
]

df_2025 = df_2025.drop(columns=["SoldasVacantFlag"])
```

```python
df_2026 = pd.read_csv("data/PropertyValue/armslengthsales_2026_valid_20261019.csv")

print("2026 Property Sales Data:")
print(df_2026.head)
# Keep only rows where SoldasVacantFlag is not equal to 1 and drop the SoldasVacantFlag column
df_2026 = df_2026[
    df_2026["SoldasVacantFlag"].astype(str).str.strip() != "1"
]

df_2026 = df_2026.drop(columns=["SoldasVacantFlag"])
```

```python
df_2019_to_2026 = pd.concat([df_2019.assign(Year=2019), df_2020.assign(Year=2020), df_2021.assign(Year=2021), df_2022.assign(Year=2022), df_2023.assign(Year=2023), df_2024.assign(Year=2024), df_2025.assign(Year=2025), df_2026.assign(Year=2026)], ignore_index=True)


# Filter the DataFrame to include only residential properties with 1 unit
df_2019_to_2026 = df_2019_to_2026[
    (df_2019_to_2026["PropType"] == "Residential") &
    (pd.to_numeric(
        df_2019_to_2026["Units"], errors="coerce"
    ) == 1)
].copy()

print(df_2019_to_2026["Style"].value_counts())
# Filter out non-residential and single family properties
styles_to_exclude = [
    "Multiple Residential Bldgs",
    "Duplex O/S",
    "Duplex N/S",
    "Dplx Bungalow",
    "Duplex-Cottage",
    "Duplex Bungalow"
]

df_2019_to_2026 = df_2019_to_2026[
    ~df_2019_to_2026["Style"].isin(styles_to_exclude)
].copy()

print(df_2019_to_2026.shape)
# Drop CondoProject, PropType, and Units columns as they are no longer needed
print(df_2019_to_2026["CondoProject"].value_counts(dropna=False))
df_2019_to_2026.drop(columns=["CondoProject", "PropType", "Units"], inplace=True)
print(df_2019_to_2026.shape)
print(df_2019_to_2026.columns.tolist())
print(df_2019_to_2026.size)
print(df_2019_to_2026["Style"].value_counts())
```

```python
print(df_2019_to_2026.isnull().sum())
print(df_2019_to_2026.shape)


# For now we want to drop Half Baths(Hbath) for now because many records have a missing value in this column and later we can explore how to impute the missing values for this column.
df_2019_to_2026.drop(columns=["Hbath"], inplace=True)
print(df_2019_to_2026.shape)
```

```python
# For records with missing SalePrice, we will drop them as we cannot use them for our analysis. We will explore how to impute the missing values for this column later.
df_2019_to_2026.dropna(subset=["Sale_price"], inplace=True)
print(df_2019_to_2026.shape)
print(df_2019_to_2026.isnull().sum())

```

```python
# This will give us the number of unique values in each column of the DataFrame, which can help us understand the diversity of the data and identify potential categorical variables.
# It also helps us identify duplicate records and understand the distribution of values in each column, which can inform our data cleaning and preprocessing steps.
print(df_2019_to_2026.nunique())

#Inspect Categorical Columns
categorical_columns = ["District", "nbhd", "Style", "Extwall"]
for col in categorical_columns:
    print(f"Column: {col}")
    print(df_2019_to_2026[col].value_counts(dropna=False))
    print("\n")
```

```python
numeric_columns = ["Stories", "Year_Built", "Rooms", "FinishedSqft", "Bdrms", "Fbath", "Lotsize"]
# Convert numeric columns to numeric data type, handling commas and whitespace. Convert any non-numeric values to NaN.
for col in numeric_columns:
    df_2019_to_2026[col] = pd.to_numeric(df_2019_to_2026[col].astype("string").str.replace(",", "", regex=False).str.strip(), errors="coerce")

print(df_2019_to_2026[numeric_columns].dtypes)

for col in ["Bdrms", "Fbath", "Rooms", "Stories"]:
    print(f"Column: {col}")
    print(df_2019_to_2026[col].value_counts(dropna=False).sort_index())
    print("\n")
```

```python
suspicious_property_characteristics = df_2019_to_2026[
    (df_2019_to_2026["Bdrms"] > 15)
    | (df_2019_to_2026["Stories"] > 3)
    | (df_2019_to_2026["Rooms"] > 25)
    | (df_2019_to_2026["FinishedSqft"] <= 0)
    | (df_2019_to_2026["Lotsize"] <= 0)
]

display(suspicious_property_characteristics[["PropertyID", "Address", "Style", "Stories", "Rooms", "Bdrms", "FinishedSqft", "Lotsize"]])
```

```python
#Replace obvious outliers with NaN
df_2019_to_2026.loc[df_2019_to_2026["Bdrms"] == 2034, "Bdrms"] = np.nan
df_2019_to_2026.loc[df_2019_to_2026["Stories"] == 19, "Stories"] = np.nan
df_2019_to_2026.loc[df_2019_to_2026["FinishedSqft"] <= 0, "FinishedSqft"] = np.nan
df_2019_to_2026.loc[df_2019_to_2026["Lotsize"] <= 0, "Lotsize"] = np.nan

print(df_2019_to_2026.isnull().sum())
print(df_2019_to_2026.shape)
```

```python
print(df_2019_to_2026["Extwall"].value_counts())
# Fix the inconsistent values in the "Extwall" column by replacing them with standardized values
df_2019_to_2026["Extwall"] = df_2019_to_2026["Extwall"].replace({
    "Alum/Vynyl Siding": "Aluminum/Vinyl",
    "Masonary Frame": "Masonry/Frame"
})
print(df_2019_to_2026["Extwall"].value_counts())
```

```python
# Converting Sale_date to datetime format for easier analysis
df_2019_to_2026["Sale_date"] = pd.to_datetime(df_2019_to_2026["Sale_date"], format = "mixed", errors="coerce")
print(df_2019_to_2026["Sale_date"].dtype)


# Identify sale dates that do not match their source year
incorrect_dates = df_2019_to_2026[
    df_2019_to_2026["Sale_date"].isna() |
    (
        df_2019_to_2026["Sale_date"].dt.year
        != df_2019_to_2026["Year"]
    ) |
    (
        df_2019_to_2026["Sale_date"]
        > pd.Timestamp("2026-11-23")
    )
]

display(incorrect_dates[
    ["PropertyID", "Address", "Year", "Sale_date"]
])

```

```python
# Clean sale price column by removing any non-numeric characters and converting to float
# Convert sale prices to numeric values
df_2019_to_2026["Sale_price"] = pd.to_numeric(
    df_2019_to_2026["Sale_price"]
    .astype("string")
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip(),
    errors="coerce"
)

print("Missing sale prices:",
      df_2019_to_2026["Sale_price"].isna().sum())

print("Sale price data type:",
      df_2019_to_2026["Sale_price"].dtype)
print(df_2019_to_2026["Sale_price"].dtype)
print(df_2019_to_2026.isnull().describe())
```

```python
# Identify invalid sale prices
invalid_prices = df_2019_to_2026[
    df_2019_to_2026["Sale_price"].isna() |
    (df_2019_to_2026["Sale_price"] <= 0)
]

display(invalid_prices[
    ["PropertyID", "Address", "Sale_price"]
])
```

```python
# Remove invalid sale dates and prices
df_2019_to_2026 = df_2019_to_2026[
    (df_2019_to_2026["Sale_price"] > 0) &
    df_2019_to_2026["Sale_date"].notna() &
    (
        df_2019_to_2026["Sale_date"].dt.year
        == df_2019_to_2026["Year"]
    ) &
    (
        df_2019_to_2026["Sale_date"]
        <= pd.Timestamp("2026-11-24")
    )
].copy()

print(df_2019_to_2026.shape)
```

```python
# Identify potential duplicate transactions
duplicates = df_2019_to_2026[
    df_2019_to_2026.duplicated(
        subset=["taxkey", "Sale_date", "Sale_price"],
        keep=False
    )
]

display(duplicates[
    ["taxkey", "Address", "Sale_date", "Sale_price"]
].sort_values(["taxkey", "Sale_date"]))


# Remove duplicated transactions
df_2019_to_2026 = df_2019_to_2026.drop_duplicates(
    subset=["taxkey", "Sale_date", "Sale_price"],
    keep="first"
).copy()

print(df_2019_to_2026.shape)
```

```python
# Extract the year and month from Sale_date
df_2019_to_2026["Sale_Year"] = (
    df_2019_to_2026["Sale_date"].dt.year
)

df_2019_to_2026["Sale_Month"] = (
    df_2019_to_2026["Sale_date"].dt.month
)

# Check the new columns
display(df_2019_to_2026[
    ["Sale_date", "Sale_Year", "Sale_Month"]
].head())
```

```python
# Calculate annual housing sale statistics
annual_summary = df_2019_to_2026.groupby(
    "Sale_Year"
).agg(
    Total_Sales=("Sale_price", "count"),
    Average_Sale_Price=("Sale_price", "mean"),
    Median_Sale_Price=("Sale_price", "median"),
    Minimum_Sale_Price=("Sale_price", "min"),
    Maximum_Sale_Price=("Sale_price", "max")
)

# Round prices to two decimal places
annual_summary = annual_summary.round(2)

display(annual_summary)
```

```python
import matplotlib.pyplot as plt

# Use complete years for the annual trend
graph_data = annual_summary.loc[2019:2025]

plt.figure(figsize=(10, 6))

plt.plot(
    graph_data.index,
    graph_data["Median_Sale_Price"],
    marker="o"
)

plt.title("Milwaukee Single-Family Home Sale Prices (2019–2025)")
plt.xlabel("Year")
plt.ylabel("Median Sale Price ($)")
plt.grid(True)

plt.tight_layout()
plt.show()
```
